## Importing Libraries

```python
import glob
import json
from tqdm import tqdm
import random
import os
from groq import Groq
```

## Setup Environment

```python
GROQ_KEY = os.getenv("GROQ_API_KEY")
PATIENT_PROFILES = glob.glob("./patient_profiles/patient_*.json")
GEN_PROMPT = glob.glob("./prompts/diary_generation_prompt.txt")
SYS_PROMPT = glob.glob("./prompts/system_prompt.txt")
```

## Generating Diaries

```python
client = Groq(api_key=GROQ_KEY)

for patient in PATIENT_PROFILES:
    print("Processing patient:", patient)
    with open(patient, "r") as f, open(GEN_PROMPT, "r") as gen_prompt_file, open(SYS_PROMPT, "r") as sys_prompt_file:
        patient_data = json.load(f)
        gen_prompt = gen_prompt_file.read()
        sys_prompt = sys_prompt_file.read()
        
        print("Generating diary entries for patient:", patient)
        
        completion = client.chat.completions.create(
            model="groq/compound",
            messages=[
                {
                    "role": "system",
                    "content": sys_prompt
                },
                {
                    "role": "user",
                    "content": gen_prompt
                }
            ]
        )
        result = completion.choices[0].message.content
```
